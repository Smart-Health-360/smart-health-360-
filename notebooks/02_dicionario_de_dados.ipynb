```python
# Carregamento do dataset
import pandas as pd

df = pd.read_csv("../data/processed/kaggle_tratado.csv")

print("Dimensão do dataset:", df.shape)

df.head()
```

## Visão Geral do Dataset

Antes de construir o dicionário de dados, vamos observar a estrutura do dataset tratado.

Nesta etapa analisamos:

- quantidade de registros

- tipos de dados

```python
df.info()
```

```python
# Distribuição dos tipos de dados
import matplotlib.pyplot as plt

tipos = df.dtypes.value_counts()

plt.figure(figsize=(6,4))
tipos.plot(kind="bar")

plt.title("Distribuição dos Tipos de Dados")
plt.xlabel("Tipo de dado")
plt.ylabel("Quantidade de variáveis")

plt.show()
```
