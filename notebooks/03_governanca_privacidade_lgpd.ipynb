# Governança de Dados, LGPD e Inteligência Artificial Responsável

Este notebook apresenta as práticas de governança de dados e proteção de privacidade adotadas no projeto de Ciência de Dados voltado para a previsão de faltas em consultas médicas.

Considerando que o projeto utiliza informações relacionadas à saúde, é fundamental avaliar aspectos éticos e legais associados ao tratamento desses dados.

No contexto brasileiro, a proteção de dados pessoais é regulamentada pela Lei Geral de Proteção de Dados (LGPD).

Essa legislação estabelece princípios e diretrizes para o tratamento de dados pessoais, especialmente dados sensíveis, como informações relacionadas à saúde.

Além disso, o projeto utiliza técnicas de Machine Learning e prevê a implementação de um chatbot baseado em Inteligência Artificial. Dessa forma, também são consideradas boas práticas de desenvolvimento de IA responsável.

Neste notebook são apresentados:

- a classificação dos dados utilizados
- os princípios da LGPD aplicados
- as medidas de proteção de dados adotadas
- práticas de privacidade no modelo de Machine Learning
- práticas de proteção de dados no chatbot com IA

## Privacidade no Modelo de Machine Learning

O modelo de Machine Learning desenvolvido no projeto tem como objetivo prever a probabilidade de ausência de pacientes em consultas médicas.

Para garantir conformidade com princípios de privacidade:

- apenas variáveis relevantes foram utilizadas
- não são utilizados identificadores pessoais
- os dados são utilizados apenas de forma agregada

Entre as variáveis utilizadas estão:

- idade
- tempo de espera entre agendamento e consulta
- histórico de doenças
- recebimento de lembrete por SMS

## Proteção de Dados no Chatbot com Inteligência Artificial

O projeto prevê a implementação de um chatbot inteligente para apoiar a comunicação com pacientes e reduzir faltas em consultas.

Para garantir conformidade com a LGPD foram consideradas as seguintes práticas:

- o chatbot não solicita dados pessoais sensíveis
- não há armazenamento permanente de conversas
- as respostas são baseadas em probabilidades geradas pelo modelo
- o sistema utiliza dados anonimizados

Essas medidas reduzem riscos de exposição de informações médicas e seguem princípios de privacidade desde o design da solução.
